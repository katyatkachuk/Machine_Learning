1. Задана функція залежності оцінки від кількості годин, витрачених на навчання: $ f(x) = 10x - 5 $.
  1. З допомогою Python обчисліть похідну цієї функції.
  2. Використовуйте бібліотеку `matplotlib` для побудови графіків функції та її похідної в діапазоні значень від 0 до 10 годин. Проаналізуйте, як зміна кількості годин вивчення впливає на оцінку.



```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0, 10, 100)

f = 10 * x - 5
df = lambda x: 10

# Побудова графіка функції та її похідної

plt.figure(figsize=(10,5))
plt.plot(x, f, label='f(x) = 10 * x - 5', color='blue')
plt.axhline(y=df(0), color='red', linestyle='--', label="f'(x) = 10") # plotting the derivative as a horizontal line
plt.title("Графік функції та її похідної")
plt.xlabel("Години")
plt.ylabel("Оцінка")
plt.legend()
plt.grid(True)

plt.show()
```

**Спостереження**

Бачимо, що зі збільшенням витрачених годин на навчання, збільшується оцінка.

2. Використовуючи Python і `SymPy`, обчисліть похідні для функцій $ f(x) = \sin(x) $ та $ f(x) = \ln(x) $. Зобразіть ці функції та їхні похідні за допомогою `matplotlib`.



```python
x = sp.symbols('x')

f_1 = sp.sin(x)
f_2 = sp.ln(x)

f_1_derivative = sp.diff(f_1, x)
f_2_derivative = sp.diff(f_2, x)

print("Похідна від функції  sin(x):", f_1_derivative)
print("Похідна від функції ln(x):", f_2_derivative)
```

```python
# Plotting
# Convert symbolic functions to numerical functions

f_1_num = sp.lambdify(x, f_1, 'numpy')
f_2_num = sp.lambdify(x, f_2, 'numpy')
f_1_derivative_num = sp.lambdify(x, f_1_derivative, 'numpy')
f_2_derivative_num = sp.lambdify(x, f_2_derivative, 'numpy')

x = np.linspace(0.1, 10, 100)

# Plot sin(x) and its derivative
plt.figure(figsize =(10,5))

plt.subplot(2, 1, 1)
plt.plot(x, f_1_num(x), label = 'f(x)=sin(x)', color = "red")
plt.plot(x, f_1_derivative_num(x), label = "f'(x)=cos(x)", color = "red", linestyle = "--")
plt.title("Графік функції sin(x) та її похідної")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid(True)
         
# Plot ln(x) and its derivative
plt.subplot(2, 1, 2)
plt.plot(x, f_2_num(x), label = 'f(x)=ln(x)', color = "green")
plt.plot(x, f_2_derivative_num(x), label = "f'(x)= 1/x", color = "green", linestyle = "--")
         
plt.title("Графік функції ln(x) та її похідної")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid(True)

plt.subplots_adjust(hspace=0.5)
plt.show()
```


3. Загальні витрати компанії від обсягу виробництва задані функцією $ f(x) = x^2 - 5x + 1 $, де $ x $ — кількість вироблених одиниць. Обчисліть похідну цієї функції з допомогою SymPy та візуалізуйте значення функції і її похідної на відрізку від 0 до 100.

```python
x = sp.symbols('x')
f = x**2 - 5 * x + 1
f_derivative = sp.diff(f, x)
print("Похідна від функції  x**2 - 5x + :", f_derivative)
```

```python
f_num = sp.lambdify(x, f, 'numpy')
f_derivative_num = sp.lambdify(x, f_derivative, 'numpy')

x = np.linspace(0, 100, 50)

plt.figure(figsize =(10, 5))
plt.plot(x, f_num(x), label = 'f(x) = x**2 - 5 * x + 1', color = "purple")
plt.plot(x, f_derivative_num(x), label = "f'(x) = 2*x - 5", color = "purple", linestyle = "-.")

plt.title("Графік функції f = x**2 - 5 * x + 1 та її похідної")
plt.xlabel("Кількість вироблених одиниць")
plt.ylabel("Витрати компанії")
plt.legend()
plt.grid(True)

plt.show()
```

 **Спостереження**
 
 Таким чином, чим більше товарів виробляється на виробництві - тим більші витрати компанії.
